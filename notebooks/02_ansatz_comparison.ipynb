# Optimiser Comparison

Compare COBYLA, SPSA, and L-BFGS-B on the same H2 problem.

```python
# If running in Google Colab, uncomment the line below to install vqe-bench
# !pip install -e .

from vqe_bench.hamiltonians import build_molecule_hamiltonian
from vqe_bench.ansatze import HardwareEfficientAnsatz
from vqe_bench.optimizers import CobylaOptimizer, LBFGSBOptimizer, SPSAOptimizer
from vqe_bench.vqe import run_vqe

ham = build_molecule_hamiltonian(
    "H2", "sto-3g", [("H", (0, 0, 0)), ("H", (0, 0, 0.735))],
    mapping="jordan_wigner",
)
opts = {
    "COBYLA": CobylaOptimizer(maxiter=300),
    "L-BFGS-B": LBFGSBOptimizer(maxiter=300),
    "SPSA": SPSAOptimizer(maxiter=300),
}

for name, opt in opts.items():
    ans = HardwareEfficientAnsatz(ham["qubit_op"].num_qubits, depth=1)
    res = run_vqe(ham, ans, opt, seed=42)
    print(f"{name:10s}  E = {res.optimal_energy:.6f}  |ΔE| = {res.error_vs_fci:.2e}  nfev = {res.num_function_evaluations}")
```
