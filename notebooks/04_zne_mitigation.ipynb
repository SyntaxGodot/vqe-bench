# Zero-Noise Extrapolation Walkthrough

Demonstrates the ZNE mitigation pipeline under depolarising noise.

```python
# If running in Google Colab, uncomment the line below to install vqe-bench
# !pip install -e .

from vqe_bench.mitigation import zne_extrapolate
import numpy as np

factors = [1.0, 3.0, 5.0]
true_energy = -1.1373
noisy = [true_energy + 0.01 * (f - 1) for f in factors]
print("Noisy energies:", noisy)

for method in ["linear", "richardson", "exponential"]:
    mitigated = zne_extrapolate(factors, noisy, method=method)
    print(f"{method:12s} → {mitigated:.6f}  (|err| = {abs(mitigated - true_energy):.2e})")
```
