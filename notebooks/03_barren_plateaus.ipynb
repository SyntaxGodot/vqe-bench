# Barren Plateaus

Gradient variance vs. qubit count for global and local cost functions.

```python
# If running in Google Colab, uncomment the line below to install vqe-bench
# !pip install -e .

import numpy as np
import matplotlib.pyplot as plt
from vqe_bench.ansatze import HardwareEfficientAnsatz
from vqe_bench.diagnostics import gradient_variance
from vqe_bench.hamiltonians import build_spin_hamiltonian

ns = [2, 3, 4, 5, 6]
vars = []

for n in ns:
    ham = build_spin_hamiltonian("tfim", n_qubits=n, j_coupling=1.0, h_field=1.0)
    def builder(nq, d):
        return HardwareEfficientAnsatz(nq, depth=d, entanglement="linear").circuit
    vars.append(gradient_variance(builder, ham["qubit_op"], n, depth=1, num_samples=30))

plt.figure(dpi=120)
plt.semilogy(ns, vars, "o-")
plt.xlabel("Qubits")
plt.ylabel("Var[∂C/∂θ]")
plt.title("Barren plateau (global cost)")
plt.show()
```
