# H2 Bond-Length Scan

In this tutorial, we scan the H–H bond length and compare the VQE (UCCSD) energy against the exact FCI reference.

```python
from vqe_bench.hamiltonians import build_molecule_hamiltonian
from vqe_bench.ansatze import UCCSDAnsatz
from vqe_bench.optimizers import CobylaOptimizer
from vqe_bench.vqe import run_vqe
import numpy as np
import matplotlib.pyplot as plt

bond_lengths = np.linspace(0.5, 2.5, 9)
vqe_energies, fci_energies = [], []

for r in bond_lengths:
    ham = build_molecule_hamiltonian(
        "H2", "sto-3g", [("H", (0, 0, 0)), ("H", (0, 0, float(r)))],
        mapping="parity", tapering=True,
    )
    ans = UCCSDAnsatz(ham)
    res = run_vqe(ham, ans, CobylaOptimizer(maxiter=300), seed=42)
    vqe_energies.append(res.optimal_energy)
    fci_energies.append(ham["fci_energy"])

plt.figure(dpi=120)
plt.plot(bond_lengths, fci_energies, "k-", label="FCI")
plt.plot(bond_lengths, vqe_energies, "ro--", label="VQE (UCCSD)")
plt.xlabel("Bond length (Å)")
plt.ylabel("Energy (Ha)")
plt.legend()
plt.show()
```
